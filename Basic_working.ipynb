```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial import distance
from sklearn.metrics.pairwise import cosine_similarity
```

**Cosine Similarity Working**

```python
vec1=np.array([1,2,3])
vec2=np.array([1,2,1])
```

```python
dot_product=np.dot(vec1,vec2)
dot_product
```

```python
#Magnitude
mag_1=np.linalg.norm(vec1)
mag_2=np.linalg.norm(vec2)
```

```python
numpy_similarity = dot_product/(mag_1*mag_2)
skelarn_similarity = cosine_similarity(vec1.reshape(1,-1),vec2.reshape(1,-1))
#We used the reshape here because it takes the 2d array as the parameter but by default the vectors were in 1D
```

```python
skelarn_similarity[0][0]
```

```python
numpy_distance = 1 - numpy_similarity
numpy_distance
```

```python
sklearn_distance = 1 - skelarn_similarity
sklearn_distance
```

```python
dist = distance.cosine(vec1, vec2)
sim = 1 - dist

# 3. Create a 3D Plot
fig = plt.figure(figsize=(7, 7))
ax = fig.add_subplot(projection='3d')

# Draw lines from origin (0,0,0) to our vector coordinates
ax.plot([0, vec1[0]], [0, vec1[1]], [0, vec1[2]], color='blue', linewidth=3, label='Vector 1 [1, 2, 3]')
ax.plot([0, vec2[0]], [0, vec2[1]], [0, vec2[2]], color='red', linewidth=3, label='Vector 2 [1, 2, 1]')

# Set axis limits
ax.set_xlim([0, 3])
ax.set_ylim([0, 3])
ax.set_zlim([0, 3])

# Add labels and details
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
ax.set_title(f"Similarity: {sim:.2f} | Distance: {dist:.2f}", fontsize=14)
ax.legend()

plt.show()
```
